```python
df['Location']=df['Location'].str.strip().replace(['UNKNOWN', 'ERROR', 'NA'], 'Unknown').fillna('Unknown') # Removing leading and trailing whitespace and standardizing 'Unknown' values in the 'Location' column
```

```python
df['Transaction Date']=pd.to_datetime(df['Transaction Date'], errors='coerce') # Converting 'Transaction Date' to datetime format, coercing errors to NaT for invalid dates
```

```python
df['Transaction Date']=df['Transaction Date'].fillna(pd.Timestamp('1900-01-01')) # Filling missing 'Transaction Date' values with a placeholder date (January 1, 1900)
```

```python
df['Payment Method'].value_counts(dropna=False) # Checking unique values and their frequencies in the 'Payment Method' column, including missing values
```

```python
df['Payment Method']=df['Payment Method'].str.strip().replace(['UNKNOWN', 'ERROR', 'NA'], 'Unknown').fillna('Unknown') # Removing leading and trailing whitespace and standardizing 'Unknown' values in the 'Payment Method' column
```

```python
df['Item'].value_counts(dropna=False)
```

```python
df['Item']=df['Item'].str.strip().replace(['UNKNOWN', 'ERROR', 'NA'], 'Unknown').fillna('Unknown') # Removing leading and trailing whitespace and standardizing 'Unknown' values in the 'Item' column
```

```python
df['Quantity'].value_counts(dropna=False) # Checking unique values and their frequencies in the 'Location' column, including missing values
```

```python
df['Quantity']=df['Quantity'].replace(['UNKNOWN', 'ERROR'], np.nan) # Replacing invalid entries in the 'Quantity' column with NaN
```

```python
df['Price Per Unit'].value_counts(dropna=False)
```

```python
df['Price Per Unit']=df['Price Per Unit'].replace(['UNKNOWN', 'ERROR'], np.nan)
```

```python
df['Total Spent'].value_counts(dropna=False)
```

```python
df['Total Spent']=df['Total Spent'].replace(['UNKNOWN', 'ERROR'], np.nan)
```

```python
df['Transaction Date'].value_counts(dropna=False)
```

```python
df['Transaction Date'] = (df['Transaction Date'].replace(['UNKNOWN', 'ERROR'], np.nan))
```

```python
df[df['Total Spent'].isna()] #Checking if there are any rows where 'Total Spent' is missing to see if they can be calculated from 'Quantity' and 'Price Per Unit'
```

```python
df['Price Per Unit']=pd.to_numeric(df['Price Per Unit'], errors='coerce') # Converting 'Price Per Unit' to numeric type to enable calculations
df['Quantity']=pd.to_numeric(df['Quantity'], errors='coerce' ) # Converting 'Quantity' to numeric type to enable calculations
df['Total Spent']=pd.to_numeric(df['Total Spent'], errors='coerce') # Converting 'Total Spent' to numeric type to enable calculations
```

```python
df[['Item','Price Per Unit']].value_counts(dropna=False) # Checking unique combinations of 'Item' and 'Price Per Unit' to identify potential inconsistencies in pricing for the same item
```

```python
df['Total Spent']=df['Total Spent'].fillna(df['Quantity']*df['Price Per Unit']) #Calculating 'Total Spent' for rows where it is missing using 'Quantity' and 'Price Per Unit'
```

```python
df[(df['Total Spent'].notna())&(df['Quantity'].isna() | df['Price Per Unit'].isna())] # Checking if there are any rows where 'Total Spent' is present but 'Quantity' or 'Price Per Unit' is missing, which would indicate that 'Total Spent' cannot be calculated for those rows
```

```python
df['Quantity']=df['Quantity'].fillna(df['Total Spent']/df['Price Per Unit']) # Calculating 'Quantity' for rows where it is missing using 'Total Spent' and 'Price Per Unit'
```

```python
df['Price Per Unit']=df['Price Per Unit'].fillna(df['Total Spent']/df['Quantity']) # Calculating 'Price Per Unit' for rows where it is missing using 'Total Spent' and 'Quantity'
```

```python
prices={'Coffee': 2.0,
        'Tea': 1.5,
        'Smoothie': 4.0,
        'Sandwich': 4.0,
        'Cookie': 1.0,
        'Salad': 5.0,
        'Cake': 3.0,
        'Juice': 3.0}    # Defining a dictionary with the most common price for each item based on the value counts
df['Price Per Unit']=df['Price Per Unit'].fillna(df['Item'].map(prices)) # Filling missing 'Price Per Unit' values based on the most common price for each item using the defined dictionary
```

```python
df[['Item','Price Per Unit']].value_counts(dropna=False) 
```

```python
item_map={2.0: 'Coffee',
          1.5: 'Tea',
          1.0: 'Cookie',
          5.0: 'Salad'}   # Defining a dictionary for products with unique prices

df.loc[df['Item']=="Unknown","Item"]=df.loc[df['Item']=="Unknown","Price Per Unit"].map(item_map) # Filling missing 'Item' values based on the unique price using the defined dictionary
```

```python
df[['Item','Price Per Unit']].value_counts(dropna=False) 
```

```python
df['Quantity']=df['Quantity'].fillna(df['Total Spent']/df['Price Per Unit']) # Calculating 'Quantity' for rows where it is missing using 'Total Spent' and 'Price Per Unit'
```

```python
df[df['Transaction ID'].duplicated(keep=False)] # Checking for duplicate 'Transaction ID' values to identify potential duplicate rows in the dataset
```

```python
df[df['Item'].isna()] #Checking what left 
```

```python
df.loc[(df['Item'].isna()) & (df['Price Per Unit'] == 4.0), 'Item'] = 'Sandwich/Smothie'
df.loc[(df['Item'].isna())&(df['Price Per Unit']==3.0),'Item']='Juice/Cake'# Filling missing 'Item' values for rows where 'Price Per Unit' is 4.0 with 'Sandwich', which is the most common item with that price
```

```python
df['Item'].value_counts(dropna=False) # Checking unique values and their frequencies in the 'Item' column after cleaning to ensure consistency
```

```python
df=df.dropna(subset=['Item','Quantity','Total Spent']) # Dropping rows where 'Item' is still missing after attempts to fill it based on price, as these rows cannot be reliably imputed
```
